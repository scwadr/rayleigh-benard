```python
import numpy as np

print("Loading preprocessed arrays...")
load_path = '/storage/scratch1/2/ashelestov3/preprocessed_4096_data.npz'
save_path = '/storage/scratch1/2/ashelestov3/preprocessed_4096_data_downsampled.npz'
data = np.load(load_path)

U_raw = data['U']
V_raw = data['V']
P_raw = data['P']
B_raw = data['B']
W_raw = data['W']
x_raw = data['x']
y_raw = data['y']
t_raw = data['t']

# Grid Parameters
size_x, size_y = U_raw.shape[0], U_raw.shape[1]
Nt = U_raw.shape[2]
Lx, Ly = 4.0, 1.0

# Calculate grid spacing (dx) for filter sizing
dx = x_raw[1] - x_raw[0]
print(f"Grid spacing dx detected as: {dx}")
print(f"Original time steps: {Nt}")

# --- NEW CODE: Downsample and Save ---

# Define the downsampling rate (1 in every 30)
step = 30

print(f"Downsampling data (taking 1 every {step} timesteps)...")

# Apply slicing [start:stop:step]. 
# Assuming shape is (x, y, t), we slice the 3rd dimension (index 2).
U_save = U_raw[:, :, ::step]
V_save = V_raw[:, :, ::step]
P_save = P_raw[:, :, ::step]
B_save = B_raw[:, :, ::step]
W_save = W_raw[:, :, ::step]
t_save = t_raw[::step]

print(t_save[1]-t_save[0])

# Note: x and y are spatial coordinates and usually do not change with time,
# so we save them as-is.
x_save = x_raw
y_save = y_raw

print(f"New time steps: {U_save.shape[2]}")
print(f"Saving to {save_path}...")

np.savez(
    save_path, 
    U=U_save, 
    V=V_save, 
    P=P_save, 
    B=B_save, 
    W=W_save, 
    x=x_save, 
    y=y_save, 
    t=t_save
)

print("Save complete.")
```
