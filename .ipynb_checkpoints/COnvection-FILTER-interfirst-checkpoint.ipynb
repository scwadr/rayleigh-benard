```python
import joblib
import h5py
import numpy as np
from utils import save, load
from library import *
import gc
from continuous_process_library_terms import *
from commons_identify_models import *
import copy
import jax
import jax.numpy as jnp
from jax_cfd.base import boundaries, grids, pressure, finite_differences as fd
import jax_cfd.base as cfd
import jax_cfd.base.grids as grids
import jax_cfd.spectral as spectral
import numpy as np
#import xarray 
#import xarray as xr
import dataclasses
import time
jax.config.update("jax_enable_x64", True)
import numpy as np
import os, joblib
from joblib import Parallel, delayed
from tqdm.auto import tqdm
from numpy.polynomial.chebyshev import chebvander

import contextlib
import joblib
from tqdm import tqdm

@contextlib.contextmanager
def tqdm_joblib(tqdm_object):
    """Context manager to patch joblib to report into tqdm progress bar given as argument"""
    class TqdmBatchCompletionCallback(joblib.parallel.BatchCompletionCallBack):
        def __call__(self, *args, **kwargs):
            tqdm_object.update(n=self.batch_size)
            return super().__call__(*args, **kwargs)

    old_batch_callback = joblib.parallel.BatchCompletionCallBack
    joblib.parallel.BatchCompletionCallBack = TqdmBatchCompletionCallback
    try:
        yield tqdm_object
    finally:
        joblib.parallel.BatchCompletionCallBack = old_batch_callback
        tqdm_object.close()
```

```python
print("Loading preprocessed arrays from .npz file...")
load_path = '/storage/scratch1/2/ashelestov3/preprocessed_4096_data.npz'
data = np.load(load_path)

U = data['U']
V = data['V']
P = data['P']
B = data['B']
W = data['W']
x_raw = data['x']
y_raw = data['y']
t = data['t']

# Original Parameters from your file
size_x = 4096
size_y = 1536
inner_steps = 1
outer_steps = 286

Lx = 4.0
Ly = 1.0
Nx = size_x
Ny = size_y
Nt = outer_steps
dx = Lx / (size_x - 1)
dy = Ly / (size_y - 1)
Lt = 1.99
dt = t[0]-t[1]
Rayleigh = 1e8
Prandtl = 1
kappa = (Rayleigh * Prandtl)**(-1/2)

print("Done loading!")
print("U shape:", U.shape)
print("P shape:", P.shape)
print(f"dx: {dx}, dt: {dt}")
```

```python
print("Setting up Grids and Matrices...")

# Parameters
delta = 0.0001
# TARGET GRID DIMENSIONS (Preserving original size as requested)
Nx_out = 4096
Ny_out = 1536 

# 1. Define Input Chebyshev nodes (Mapped to 0-1)
# Assuming input y is Chebyshev as in original file
# y_raw is likely already linspace or Chebyshev nodes in ascending order
y_nodes_in = y_raw 
perm = np.arange(Ny) # Identity mapping (do nothing)

# 2. Define Output Target Grid (Uniform)
y_out = np.linspace(0, Ly, Ny_out)

# 3. Build Projection Matrix P (Chebyshev -> Uniform)
# Note: Input is Cheb, Output is Uniform. 
# We build P such that: U_uniform = U_cheb @ P.T
def _build_P(x_in, x_out):
    # x_in: Chebyshev nodes (0 to 1)
    # x_out: Target uniform nodes (0 to 1)
    Nyc = x_in.size
    # Map to [-1, 1] for chebvander
    t_in = 2.0 * x_in - 1.0
    t_out = 2.0 * x_out - 1.0
    # Create Vandermonde matrices
    V_in = chebvander(t_in, Nyc - 1)
    V_out = chebvander(t_out, Nyc - 1)
    # Solve for projection weights: V_out = P @ V_in  =>  P = V_out @ inv(V_in)
    # We transpose because we apply as: Data @ P.T
    # V_in.T @ C = V_out.T  => C = P.T
    return np.linalg.solve(V_in.T, V_out.T).T

P_interp = _build_P(y_nodes_in, y_out)

# 4. Build Smoothing Matrix S (On Target Uniform Grid)
# Now that data is uniform, S is a Gaussian convolution matrix
def _build_S(y_grid, delta, Ly):
    N = y_grid.size
    delta_scaled = delta / Ly
    yk, yj = y_grid[None, :], y_grid[:, None]
    G = np.exp(-6.0 * ((yk - yj)**2) / (delta_scaled**2))
    # Riemann sum weights for uniform grid
    dy = y_grid[1] - y_grid[0]
    w = np.full_like(y_grid, dy)
    w[0] *= 0.5; w[-1] *= 0.5 # Trapezoidal
    WG = w[None, :] * G
    S = WG / WG.sum(axis=1, keepdims=True)
    return S

S_filter = _build_S(y_out, delta, Ly)

# 5. Build X-Filter (Fourier)
def _build_Gx(n, Lx, delta):
    kx = np.fft.rfftfreq(n, d=Lx/n) * 2 * np.pi
    return np.exp(-kx**2 * delta**2 / 24.0)[:, None]

Gx_vec = _build_Gx(Nx_out, Lx, delta)

print("Matrices built.")
```

```python
# # Processing Loop (Chunked & Parallel)

# %%
# Output Arrays
u_filt = np.empty((Nx_out, Ny_out, Nt), dtype=U.dtype)
v_filt = np.empty((Nx_out, Ny_out, Nt), dtype=V.dtype)
p_filt = np.empty((Nx_out, Ny_out, Nt), dtype=P.dtype)
b_filt = np.empty((Nx_out, Ny_out, Nt), dtype=B.dtype)
w_filt = np.empty((Nx_out, Ny_out, Nt), dtype=W.dtype)

tau_xx = np.empty((Nx_out, Ny_out, Nt), dtype=U.dtype)
tau_xy = np.empty((Nx_out, Ny_out, Nt), dtype=U.dtype)
tau_yy = np.empty((Nx_out, Ny_out, Nt), dtype=U.dtype)

def chunked_matmul(Matrix, Field, n_chunks=4):
    # Field shape: (Nx, Ny)
    # Matrix shape: (Ny_new, Ny_old)
    # Result = Field @ Matrix.T -> (Nx, Ny_new)
    # We transpose Field to (Ny, Nx) to apply Matrix on rows
    F_T = Field.T
    Res_T = np.zeros((Matrix.shape[0], F_T.shape[1]), dtype=Field.dtype)
    
    chunk = (Matrix.shape[0] + n_chunks - 1) // n_chunks
    for i in range(n_chunks):
        s = i * chunk
        e = min(s + chunk, Matrix.shape[0])
        if s >= e: break
        Res_T[s:e] = Matrix[s:e] @ F_T
    return Res_T.T

def process_timestep(ti):
    # 1. Load Raw Data (sorted by y)
    u_i = U[:, perm, ti]
    v_i = V[:, perm, ti]
    p_i = P[:, perm, ti]
    b_i = B[:, perm, ti]
    w_i = W[:, perm, ti]
    
    # 2. Interpolate Y (Cheb -> Uniform) using P_interp
    # Op: Data_New = Data_Old @ P.T
    u_int = chunked_matmul(P_interp, u_i)
    v_int = chunked_matmul(P_interp, v_i)
    p_int = chunked_matmul(P_interp, p_i)
    b_int = chunked_matmul(P_interp, b_i)
    w_int = chunked_matmul(P_interp, w_i)
    
    u2_int = chunked_matmul(P_interp, u_i**2)
    v2_int = chunked_matmul(P_interp, v_i**2)
    uv_int = chunked_matmul(P_interp, u_i*v_i)
    
    # 3. Filter (Applied to Interpolated Uniform Data)
    
    # Helper: X-Filter (FFT) then Y-Filter (Matrix)
    def apply_filters(field):
        # Filter X
        ft = np.fft.rfft(field, axis=0)
        ft_filt = ft * Gx_vec
        field_x = np.fft.irfft(ft_filt, n=Nx_out, axis=0)
        # Filter Y (S_filter is on uniform grid)
        return chunked_matmul(S_filter, field_x)

    u_bar = apply_filters(u_int)
    v_bar = apply_filters(v_int)
    p_bar = apply_filters(p_int)
    b_bar = apply_filters(b_int)
    w_bar = apply_filters(w_int)
    
    uu_bar = apply_filters(u2_int)
    vv_bar = apply_filters(v2_int)
    uv_bar = apply_filters(uv_int)
    
    # 4. Stresses
    txx = uu_bar - u_bar**2
    tyy = vv_bar - v_bar**2
    txy = uv_bar - u_bar*v_bar
    
    return ti, u_bar, v_bar, p_bar, b_bar, w_bar, txx, txy, tyy

print("Starting Parallel Processing (Interpolate -> Filter)...")
N_JOBS = max(1, joblib.cpu_count() - 4)

with tqdm_joblib(tqdm(total=Nt)) as progress_bar:
    results = Parallel(n_jobs=N_JOBS)(delayed(process_timestep)(ti) for ti in range(Nt))

print("Collecting results...")
for ti, u, v, p, b, w, txx, txy, tyy in results:
    u_filt[..., ti] = u
    v_filt[..., ti] = v
    p_filt[..., ti] = p
    b_filt[..., ti] = b
    w_filt[..., ti] = w
    tau_xx[..., ti] = txx
    tau_xy[..., ti] = txy
    tau_yy[..., ti] = tyy

# Prepare for Spider
u_bar_stack = np.stack([u_filt, v_filt], axis=3)
# Tensor structure [[xx, xy], [yx, yy]]
row1 = np.stack([tau_xx, tau_xy], axis=3)
row2 = np.stack([tau_xy, tau_yy], axis=3) # Symm
tau = np.stack([row1, row2], axis=4)

print("Filtering Complete. Shapes:")
print("u_bar:", u_bar_stack.shape)
print("tau:", tau.shape)
```

```python
import numpy as np
import matplotlib.pyplot as plt
 
def compute_derivatives_and_residuals(u, v, p, b, tau, x, y, dt, kappa):
    """
    Computes terms of the momentum equations using np.gradient (central difference).
    Returns the residuals for U and V momentum equations.
    """
    print("Computing derivatives...")
    # Grid Spacing
    dx = x[1] - x[0]
    dy = y[1] - y[0]
    # --- 1. Temporal Derivative (d/dt) ---
    # axis=2 is time
    u_t = np.gradient(u, dt, axis=2)
    v_t = np.gradient(v, dt, axis=2)
    # --- 2. Spatial Derivatives (First Order) ---
    # axis=0 is X, axis=1 is Y
    u_x = np.gradient(u, dx, axis=0)
    u_y = np.gradient(u, dy, axis=1)
    v_x = np.gradient(v, dx, axis=0)
    v_y = np.gradient(v, dy, axis=1)
    p_x = np.gradient(p, dx, axis=0)
    p_y = np.gradient(p, dy, axis=1)
    # --- 3. Viscous Terms (Laplacian) ---
    u_xx = np.gradient(u_x, dx, axis=0)
    u_yy = np.gradient(u_y, dy, axis=1)
    lap_u = u_xx + u_yy
    v_xx = np.gradient(v_x, dx, axis=0)
    v_yy = np.gradient(v_y, dy, axis=1)
    lap_v = v_xx + v_yy
    # --- 4. Subgrid Stress Divergence (div Tau) ---
    # tau shape: (Nx, Ny, Nt, 2, 2) -> (xx, xy; yx, yy)
    tau_xx = tau[..., 0, 0]
    tau_xy = tau[..., 0, 1]
    tau_yy = tau[..., 1, 1]
    # div(Tau)_x = d(tau_xx)/dx + d(tau_xy)/dy
    dtxx_dx = np.gradient(tau_xx, dx, axis=0)
    dtxy_dy = np.gradient(tau_xy, dy, axis=1)
    div_tau_x = dtxx_dx + dtxy_dy
    # div(Tau)_y = d(tau_xy)/dx + d(tau_yy)/dy
    dtxy_dx = np.gradient(tau_xy, dx, axis=0)
    dtyy_dy = np.gradient(tau_yy, dy, axis=1)
    div_tau_y = dtxy_dx + dtyy_dy
 
    # --- 5. Assemble Residuals ---
    # U-Momentum: du/dt + u*du/dx + v*du/dy = -dp/dx + nu*Lap(u) - div(Tau)_x
    # Residual = LHS - RHS
    # Note: If tau is defined as (Mean(uu) - Mean(u)Mean(u)), it sits on the RHS with a negative sign
    # as usually written in LES: d_t u_bar + div(u_bar u_bar) = ... - div(tau)
    adv_u = (u * u_x) + (v * u_y)
    # Res_U = Time + Advection + Pressure - Diffusion + Div(Tau)
    # Signs depend on how you move terms. Here we calculate:
    # Error = (Time + Advec) - (Pressure + Diffusion + Stress)
    # Check signs: Pressure is usually -dp/dx. Diffusion is +nu*Lap.
    # Standard Form: du/dt + u.grad(u) = -grad(p) + nu*Lap(u) - div(tau)
    # So: Res = (du/dt + u.grad(u)) - (-grad(p) + nu*Lap(u) - div(tau))
    Res_U = (u_t + adv_u) - (-p_x + kappa * lap_u - div_tau_x)
    # V-Momentum: dv/dt + u*dv/dx + v*dv/dy = -dp/dy + nu*Lap(v) + Buoyancy - div(Tau)_y
    # Buoyancy term usually +b (or T) in vertical direction
    adv_v = (u * v_x) + (v * v_y)
    Res_V = (v_t + adv_v) - (-p_y + kappa * lap_v + b - div_tau_y)
    return Res_U, Res_V
 
# --- EXECUTION ---

print("Calculating Residuals...")

# Unpack your variables
# u_bar shape is (Nx, Ny, Nt, 2)
_u = u_bar[..., 0]
_v = u_bar[..., 1]
_p = p_filt
_b = b_filt

# FIX: Use 'xp' and 'yp' (the actual filtered coordinates) instead of 'x' and 'y'
# 'x' and 'y' might have been overwritten by the SRDataset cell
res_u, res_v = compute_derivatives_and_residuals(_u, _v, _p, _b, tau, xp, yp, dt, kappa)

print("Calculation Complete.")
print(f"Residual U - Mean: {np.mean(np.abs(res_u)):.6f}, Max: {np.max(np.abs(res_u)):.6f}")
print(f"Residual V - Mean: {np.mean(np.abs(res_v)):.6f}, Max: {np.max(np.abs(res_v)):.6f}")

# --- PLOTTING ---
t_idx = 50

fig, ax = plt.subplots(1, 2, figsize=(14, 6))

im1 = ax[0].imshow(res_u[:, :, t_idx].T, origin='lower', cmap='seismic', aspect='auto')
ax[0].set_title(f"U-Momentum Residual (t={t_idx})")
plt.colorbar(im1, ax=ax[0])

im2 = ax[1].imshow(res_v[:, :, t_idx].T, origin='lower', cmap='seismic', aspect='auto')
ax[1].set_title(f"V-Momentum Residual (t={t_idx})")
plt.colorbar(im2, ax=ax[1])

plt.show()
 
# --- PLOTTING ---
# Visualize the error at a specific timestep
t_idx = 50
 
fig, ax = plt.subplots(1, 2, figsize=(14, 6))
 
im1 = ax[0].imshow(res_u[:, :, t_idx].T, origin='lower', cmap='seismic', aspect='auto')
ax[0].set_title(f"U-Momentum Residual (t={t_idx})")
plt.colorbar(im1, ax=ax[0])
 
im2 = ax[1].imshow(res_v[:, :, t_idx].T, origin='lower', cmap='seismic', aspect='auto')
ax[1].set_title(f"V-Momentum Residual (t={t_idx})")
plt.colorbar(im2, ax=ax[1])
 
plt.show()
```
