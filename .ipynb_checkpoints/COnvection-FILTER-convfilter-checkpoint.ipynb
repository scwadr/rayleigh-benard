```python
def _sigma_from_delta(delta):
    return delta / np.sqrt(12.0)
 
def _next_fast_len(n: int) -> int:
    return 1 << (n - 1).bit_length()
 
def _gaussian_kernel_1d_uniform(dx, delta, alpha):
    sigma = _sigma_from_delta(delta)
    R = alpha * delta
 
    r = int(np.ceil(R / dx))                # radius in grid points
    xs = (np.arange(-r, r + 1) * dx)        # symmetric, odd length
    k = np.exp(-0.5 * (xs / sigma) ** 2)
 
    # hard cutoff at |x|<=R (matches your mask logic)
    k[np.abs(xs) > R] = 0.0
 
    s = k.sum()
    if s == 0.0:
        k[r] = 1.0
        s = 1.0
    k /= s
    return k, r  # kernel and integer radius

def _fft_convolve1d_same_axis(arr, kernel, axis):
    import builtins
    arr = np.asarray(arr)
    k = np.asarray(kernel)
    m = k.size
    if m == 1:
        return arr.copy()
    n = arr.shape[axis]
    full_len = n + m - 1
    L = _next_fast_len(full_len)
    
    K = np.fft.rfft(k, n=L)
    A = np.fft.rfft(arr, n=L, axis=axis)
    
    # --- FIX: Correct broadcasting for any axis ---
    shape = [1] * A.ndim
    shape[axis] = K.shape[0]
    C = A * K.reshape(shape)
    # ----------------------------------------------

    c_full = np.fft.irfft(C, n=L, axis=axis)
    
    # Slice to keep original length logic
    sl = [builtins.slice(None)] * c_full.ndim
    sl[axis] = builtins.slice(0, full_len)
    c_full = c_full[tuple(sl)]
    
    start = (m - 1) // 2
    end = start + n
    sl[axis] = builtins.slice(start, end)
    
    return c_full[tuple(sl)]
 
def trim_edges_2d_uniform(field, x, y, trim_dist):
    x = np.asarray(x); y = np.asarray(y)
    dx = float(x[1] - x[0]) if x.size > 1 else 1.0
    dy = float(y[1] - y[0]) if y.size > 1 else 1.0
    rx = int(np.ceil(trim_dist / dx))
    ry = int(np.ceil(trim_dist / dy))
    if rx == 0 and ry == 0:
        return field, x, y
    trimmed = field[ry: field.shape[0] - ry, rx: field.shape[1] - rx]
    x_new = x[rx: x.size - rx]
    y_new = y[ry: y.size - ry]
    return trimmed, x_new, y_new

def gaussian_filter_2d_uniform_finite_support(field, x, y, delta, alpha):
    field = np.asarray(field)
    x = np.asarray(x); y = np.asarray(y)
    
    # NOTE: The provided code expects field dimensions to match x and y sizes.
    # Data is (Nx, Ny) -> (4096, 1536).
    # x size 4096, y size 1536.
    # Check consistency and apply correctly.
    if field.shape[0] != x.size:
        # If mismatch, try transposing locally or ensure x corresponds to axis 0
        if field.shape[1] == x.size: 
             field = field.T # Flip to match
        else:
             raise ValueError(f"field shape {field.shape} does not match x={x.size}, y={y.size}")

    dx = float(x[1] - x[0]) if x.size > 1 else 1.0
    dy = float(y[1] - y[0]) if y.size > 1 else 1.0
    
    kx, _ = _gaussian_kernel_1d_uniform(dx, delta, alpha)
    ky, _ = _gaussian_kernel_1d_uniform(dy, delta, alpha)
    
    # Filter Axis 0 (x)
    tmp = _fft_convolve1d_same_axis(field, kx, axis=0) 
    # Filter Axis 1 (y)
    out = _fft_convolve1d_same_axis(tmp,  ky, axis=1) 
    
    R = alpha * delta
    # We do NOT trim inside the loop to keep array sizes consistent for stacking. 
    # Trim should happen later if desired.
    # out_trim, x_trim, y_trim = trim_edges_2d_uniform(out, x, y, trim_dist=R)
    return out
```

```python
# # Main Pipeline (Interpolate -> Filter)

# %%
delta = 0.005
alpha = 3

# 1. Setup Interpolation (Cheb -> Uniform)
# Target Grid
x_uni = np.linspace(0, Lx, Nx)
y_uni = np.linspace(0, Ly, Ny)

# Use Matrix Interpolation logic to get to Uniform Grid first
# Use the known sorted grid directly
y_nodes_in = np.linspace(0, 1, Ny) # Or load the actual y-grid from the file if non-uniform
# If using Chebyshev nodes in Dedalus v3, they are:
# y_nodes_in = -np.cos(np.pi * np.arange(Ny) / (Ny - 1)) * 0.5 + 0.5

perm = np.arange(Ny) # Identity mapping (0, 1, 2... N)
y_nodes_sorted = y_nodes_in # Already sorted

def _build_P(x_in, x_out):
    t_in = 2.0 * x_in - 1.0
    t_out = 2.0 * x_out - 1.0
    V_in = chebvander(t_in, x_in.size - 1)
    V_out = chebvander(t_out, x_in.size - 1)
    return np.linalg.solve(V_in.T, V_out.T).T

P_interp = _build_P(y_nodes_sorted, y_uni)

# Output containers
u_filt = np.zeros_like(U)
v_filt = np.zeros_like(V)
b_filt = np.zeros_like(B)
p_filt = np.zeros_like(P)
w_filt = np.zeros_like(W)
tau_xx = np.zeros_like(U)
tau_xy = np.zeros_like(U)
tau_yy = np.zeros_like(U)

def process_step(ti):
    # 1. Load & Sort (using perm)
    u_i = U[:, perm, ti]; v_i = V[:, perm, ti]
    b_i = B[:, perm, ti]; p_i = P[:, perm, ti]; w_i = W[:, perm, ti]
    u2_i = u_i**2
    v2_i = v_i**2
    uv_i = u_i * v_i
    
    # 2. Bypass Interpolation (Directly use sorted data)
    u_uni = u_i; v_uni = v_i; b_uni = b_i; p_uni = p_i; w_uni = w_i
    
    # 3. Filter using Gaussian Kernel
    u_bar = gaussian_filter_2d_uniform_finite_support(u_uni, x_uni, y_uni, delta, alpha)
    v_bar = gaussian_filter_2d_uniform_finite_support(v_uni, x_uni, y_uni, delta, alpha)
    b_bar = gaussian_filter_2d_uniform_finite_support(b_uni, x_uni, y_uni, delta, alpha)
    p_bar = gaussian_filter_2d_uniform_finite_support(p_uni, x_uni, y_uni, delta, alpha)
    w_bar = gaussian_filter_2d_uniform_finite_support(w_uni, x_uni, y_uni, delta, alpha)
    uu_bar = gaussian_filter_2d_uniform_finite_support(u2_i, x_uni, y_uni, delta, alpha)
    vv_bar = gaussian_filter_2d_uniform_finite_support(v2_i, x_uni, y_uni, delta, alpha)
    uv_bar = gaussian_filter_2d_uniform_finite_support(uv_i, x_uni, y_uni, delta, alpha)
    # 4. Stresses
    txx = uu_bar - u_bar**2
    tyy = vv_bar - v_bar**2
    txy = uv_bar - u_bar*v_bar
    
    return ti, u_bar, v_bar, b_bar, p_bar, w_bar, txx, txy, tyy

print("Processing...")
# Reduced n_jobs to avoid OOM with large grid
results = Parallel(n_jobs=64)(delayed(process_step)(ti) for ti in tqdm(range(Nt)))

for ti, u, v, b, p, w, txx, txy, tyy in results:
    u_filt[..., ti] = u
    v_filt[..., ti] = v
    b_filt[..., ti] = b
    p_filt[..., ti] = p
    w_filt[..., ti] = w
    tau_xx[..., ti] = txx
    tau_xy[..., ti] = txy
    tau_yy[..., ti] = tyy

# Stack
u_bar_stack = np.stack([u_filt, v_filt], axis=3)
row1 = np.stack([tau_xx, tau_xy], axis=3)
row2 = np.stack([tau_xy, tau_yy], axis=3)
tau = np.stack([row1, row2], axis=4)

print("Prompt Method Pipeline Complete.")
```

```python
import numpy as np
import matplotlib.pyplot as plt
 
def compute_derivatives_and_residuals(u, v, p, b, tau, x, y, dt, kappa):
    """
    Computes terms of the momentum equations using np.gradient (central difference).
    Returns the residuals for U and V momentum equations.
    """
    print("Computing derivatives...")
    # Grid Spacing
    dx = x[1] - x[0]
    dy = y[1] - y[0]
    # --- 1. Temporal Derivative (d/dt) ---
    # axis=2 is time
    u_t = np.gradient(u, dt, axis=2)
    v_t = np.gradient(v, dt, axis=2)
    # --- 2. Spatial Derivatives (First Order) ---
    # axis=0 is X, axis=1 is Y
    u_x = np.gradient(u, dx, axis=0)
    u_y = np.gradient(u, dy, axis=1)
    v_x = np.gradient(v, dx, axis=0)
    v_y = np.gradient(v, dy, axis=1)
    p_x = np.gradient(p, dx, axis=0)
    p_y = np.gradient(p, dy, axis=1)
    # --- 3. Viscous Terms (Laplacian) ---
    u_xx = np.gradient(u_x, dx, axis=0)
    u_yy = np.gradient(u_y, dy, axis=1)
    lap_u = u_xx + u_yy
    v_xx = np.gradient(v_x, dx, axis=0)
    v_yy = np.gradient(v_y, dy, axis=1)
    lap_v = v_xx + v_yy
    # --- 4. Subgrid Stress Divergence (div Tau) ---
    # tau shape: (Nx, Ny, Nt, 2, 2) -> (xx, xy; yx, yy)
    tau_xx = tau[..., 0, 0]
    tau_xy = tau[..., 0, 1]
    tau_yy = tau[..., 1, 1]
    # div(Tau)_x = d(tau_xx)/dx + d(tau_xy)/dy
    dtxx_dx = np.gradient(tau_xx, dx, axis=0)
    dtxy_dy = np.gradient(tau_xy, dy, axis=1)
    div_tau_x = dtxx_dx + dtxy_dy
    # div(Tau)_y = d(tau_xy)/dx + d(tau_yy)/dy
    dtxy_dx = np.gradient(tau_xy, dx, axis=0)
    dtyy_dy = np.gradient(tau_yy, dy, axis=1)
    div_tau_y = dtxy_dx + dtyy_dy
 
    # --- 5. Assemble Residuals ---
    # U-Momentum: du/dt + u*du/dx + v*du/dy = -dp/dx + nu*Lap(u) - div(Tau)_x
    # Residual = LHS - RHS
    # Note: If tau is defined as (Mean(uu) - Mean(u)Mean(u)), it sits on the RHS with a negative sign
    # as usually written in LES: d_t u_bar + div(u_bar u_bar) = ... - div(tau)
    adv_u = (u * u_x) + (v * u_y)
    # Res_U = Time + Advection + Pressure - Diffusion + Div(Tau)
    # Signs depend on how you move terms. Here we calculate:
    # Error = (Time + Advec) - (Pressure + Diffusion + Stress)
    # Check signs: Pressure is usually -dp/dx. Diffusion is +nu*Lap.
    # Standard Form: du/dt + u.grad(u) = -grad(p) + nu*Lap(u) - div(tau)
    # So: Res = (du/dt + u.grad(u)) - (-grad(p) + nu*Lap(u) - div(tau))
    Res_U = (u_t + adv_u) - (-p_x + kappa * lap_u - div_tau_x)
    # V-Momentum: dv/dt + u*dv/dx + v*dv/dy = -dp/dy + nu*Lap(v) + Buoyancy - div(Tau)_y
    # Buoyancy term usually +b (or T) in vertical direction
    adv_v = (u * v_x) + (v * v_y)
    Res_V = (v_t + adv_v) - (-p_y + kappa * lap_v + b - div_tau_y)
    return Res_U, Res_V
 
# --- EXECUTION ---

print("Calculating Residuals...")

# Unpack your variables
# u_bar shape is (Nx, Ny, Nt, 2)
_u = u_bar[..., 0]
_v = u_bar[..., 1]
_p = p_filt
_b = b_filt

# FIX: Use 'xp' and 'yp' (the actual filtered coordinates) instead of 'x' and 'y'
# 'x' and 'y' might have been overwritten by the SRDataset cell
res_u, res_v = compute_derivatives_and_residuals(_u, _v, _p, _b, tau, xp, yp, dt, kappa)

print("Calculation Complete.")
print(f"Residual U - Mean: {np.mean(np.abs(res_u)):.6f}, Max: {np.max(np.abs(res_u)):.6f}")
print(f"Residual V - Mean: {np.mean(np.abs(res_v)):.6f}, Max: {np.max(np.abs(res_v)):.6f}")

# --- PLOTTING ---
t_idx = 50

fig, ax = plt.subplots(1, 2, figsize=(14, 6))

im1 = ax[0].imshow(res_u[:, :, t_idx].T, origin='lower', cmap='seismic', aspect='auto')
ax[0].set_title(f"U-Momentum Residual (t={t_idx})")
plt.colorbar(im1, ax=ax[0])

im2 = ax[1].imshow(res_v[:, :, t_idx].T, origin='lower', cmap='seismic', aspect='auto')
ax[1].set_title(f"V-Momentum Residual (t={t_idx})")
plt.colorbar(im2, ax=ax[1])

plt.show()
 
# --- PLOTTING ---
# Visualize the error at a specific timestep
t_idx = 50
 
fig, ax = plt.subplots(1, 2, figsize=(14, 6))
 
im1 = ax[0].imshow(res_u[:, :, t_idx].T, origin='lower', cmap='seismic', aspect='auto')
ax[0].set_title(f"U-Momentum Residual (t={t_idx})")
plt.colorbar(im1, ax=ax[0])
 
im2 = ax[1].imshow(res_v[:, :, t_idx].T, origin='lower', cmap='seismic', aspect='auto')
ax[1].set_title(f"V-Momentum Residual (t={t_idx})")
plt.colorbar(im2, ax=ax[1])
 
plt.show()
```
